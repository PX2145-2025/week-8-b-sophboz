```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Mass-spring parameters
m = 1.0
k = 1.0

# System of ODEs: y = [x, v]
def mass_spring(t, y):
    x, v = y
    dxdt = v
    dvdt = -(k/m) * x
    return [dxdt, dvdt]

# Initial conditions
x0 = 1.0     # initial displacement
v0 = 0.0     # initial velocity
y0 = [x0, v0]

# Time interval
t_span = (0, 20)       # solve from t=0 to t=20
t_eval = np.linspace(0, 20, 2000)

# Solve the ODE
solution = solve_ivp(mass_spring, t_span, y0, t_eval=t_eval)

# Extract solution
t = solution.t
x = solution.y[0]

# Plot
plt.figure(figsize=(10, 5))
plt.plot(t, x, label="x(t)")
plt.title("Mass–Spring System: Solution of $x'' + x = 0$")
plt.xlabel("Time (t)")
plt.ylabel("Displacement (x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# --- Derivatives function that accepts parameters explicitly ---
def mass_spring(t, y, m, k):
    x, v = y
    dxdt = v
    dvdt = -(k/m) * x
    return [dxdt, dvdt]

# Parameters
m = 1.0
k = 1.0

# Initial conditions
x0 = 1.0
v0 = 0.0
y0 = [x0, v0]

# Time range
t_span = (0, 20)
t_eval = np.linspace(0, 20, 2000)

# Solve ODE using lambda to pass m and k into the function
solution = solve_ivp(
    lambda t, y: mass_spring(t, y, m, k),
    t_span,
    y0,
    t_eval=t_eval
)

# Extract solution
t = solution.t
x = solution.y[0]

# --- Plot ---
plt.figure(figsize=(10, 5))
plt.plot(t, x, label="x(t)")
plt.title("Mass–Spring System with Parameters Passed to Derivative Function")
plt.xlabel("Time (t)")
plt.ylabel("Displacement (x)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from scipy.optimize import curve_fit

# Load data
data = np.loadtxt("spring_measurement.txt")
t_data = data[:, 0]
x_data = data[:, 1]

plt.figure(figsize=(10,5))
plt.plot(t_data, x_data, '.', markersize=3)
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.grid(True)
plt.show()

# Damped spring model
def damped_spring_solution(t, k, b):
    m = 1.0
    x0 = 0.5
    v0 = 0.0
    y0 = [x0, v0]

    def deriv(y, t):
        x, v = y
        dxdt = v
        dvdt = -(b/m)*v - (k/m)*x
        return [dxdt, dvdt]

    sol = odeint(deriv, y0, t)
    return sol[:, 0]

# Initial guess
p0 = [5.0, 0.5]
x_guess = damped_spring_solution(t_data, *p0)

plt.figure(figsize=(10,5))
plt.plot(t_data, x_data, '.', markersize=3)
plt.plot(t_data, x_guess, '-')
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.grid(True)
plt.show()

# Fit
popt, pcov = curve_fit(damped_spring_solution, t_data, x_data, p0=p0)
k_fit, b_fit = popt
print(k_fit, b_fit)

# Plot fit
x_fit = damped_spring_solution(t_data, k_fit, b_fit)

plt.figure(figsize=(10,5))
plt.plot(t_data, x_data, '.', markersize=3)
plt.plot(t_data, x_fit, '-')
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parameters
mu = 8
A = 1
w = 0.6

# ODE system
def vdp(t, Y):
    y, yp = Y
    ypp = mu*(1 - y**2)*yp - y + A*np.sin(w*t)
    return [yp, ypp]

# Time span
t0, t1 = 0, 200
t_eval = np.linspace(t0, t1, 20000)

# Initial conditions
Y0 = [0.1, 0.0]

# Solve
sol = solve_ivp(vdp, (t0, t1), Y0, t_eval=t_eval)

t = sol.t
y = sol.y[0]
yp = sol.y[1]

# y vs t
plt.figure(figsize=(10,5))
plt.plot(t, y)
plt.xlabel("t")
plt.ylabel("y")
plt.title("Driven Van der Pol Oscillator")
plt.grid(True)
plt.show()

# Phase portrait (y' vs y)
plt.figure(figsize=(6,6))
plt.plot(y, yp, linewidth=0.7)
plt.xlabel("y")
plt.ylabel("y'")
plt.title("Phase Space Plot")
plt.grid(True)
plt.show()
```
